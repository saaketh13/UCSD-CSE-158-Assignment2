```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from evaluate import evaluate_predictions
from preprocess import preprocess_splits
```

```python
train_df = pd.read_csv("../data/processed/train.csv")
val_df = pd.read_csv("../data/processed/val.csv")
test_df = pd.read_csv("../data/processed/test.csv")

X_train, y_train, X_val, y_val, X_test, y_test = preprocess_splits(train_df, val_df, test_df)

```

```python
# Baseline 0: Predicts Fit for everything
y_val_majority = np.ones_like(y_val)
evaluate_predictions(y_val, y_val_majority, model_name="Baseline 0:Predicts Fit for everything")

# Model 1: Logistic Regression w/ default weights
lr_default = LogisticRegression(max_iter=1000, random_state=42)
lr_default.fit(X_train, y_train)
y_val_lr_default = lr_default.predict(X_val)
evaluate_predictions(y_val, y_val_lr_default, model_name="Model 1: Logistic Regression w/ Default Weights")

# Model 2: Logistic Regression w/ balanced class weights 
lr_balanced = LogisticRegression(max_iter=1000, class_weight="balanced", random_state=42)
lr_balanced.fit(X_train, y_train)
y_val_lr_balanced = lr_balanced.predict(X_val)
evaluate_predictions(y_val, y_val_lr_balanced, model_name="Model 2: Logistic Regression w/ Balanced Weights")
```

### Baseline Model
##### First, we ran a baseline that ignores all features and predicts fit for every single review. We included this to set a floor. It gets 65% accuracy without doing anything. This tells us that accuracy is somewhat misleading on this dataset. Therefore, we need to evaluate using the Balanced Accuracy as our main metric. 

### Logistic Regression w/ Default Weights
##### Next, we trained a standard Logistic Regression model using our cleaned features ( size, height, hips, bra size, cup size, category) with every training example weighted equally. We wanted to see how a basic linear model handles data on its own. Because fit makes up 65% of the dataset, the model blindly guessed fit for 8277 out of 8279 of the validation reviews and got a 32% balanced accuracy rate. 

### Balanced Logistic Regression
##### To fix that, we trained the exact same Logistic Regression Model on the same features, just with class_weight="balanced". This penalizes the model more harshly when it gets a 'small' or 'large' item wrong when compared to a 'fit' item. This forces the model to care equally about all 3 classes without just blindly guessing fit. This ended up increasing our Balanced Accuracy up to 40%.
